```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel('filmes_limpo.xlsx')
df
```

```python
# Quais são os anos mais frequentes?

anos_freq = df['ano'].value_counts()

anos_mais_freq = anos_freq.nlargest(10)
anos_mais_freq.plot(kind='bar', figsize=(10,5), colormap='Set2')

plt.title('Anos mais frequentes de lançamento dos Top 250 filmes no IMDb')
plt.xlabel('Top 10 anos de lançamento')
plt.ylabel('Quantidade de filmes lançados')
plt.xticks(rotation=45)
plt.grid(axis='y')
```

```python
# Qual a distribuição de idade minima para ver os filmes?

idade = df['classificacao'].value_counts()
idade.plot(kind='bar', figsize=(10,5), colormap='Set2')

plt.grid(axis='y')
plt.title('Distribuição de classificação de idade dos Top 250 filmes no IMDb')
plt.ylabel('Quantidade de filmes com a classificação')
plt.xlabel('Classificação de idade')
plt.xticks(rotation=45)
```
