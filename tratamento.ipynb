```python
import pandas as pd
df = pd.read_excel('filmes.xlsx')

df
```

```python

df[['ano', 'duracao', 'classificacao']] = df['ano'].str.split('\n', expand=True)
display(df.head(40))
```

```python
df['titulo'] = df['titulo'].str.replace('. ', '/', n=1)
print(df.head(163))

df[['posicao', 'titulo']] = df['titulo'].str.split('/', expand=True)

df['ano'] = df['ano'].astype(int)
df.info()

df
```

```python
df.to_excel('filmes_limpo.xlsx', index=False)
```
